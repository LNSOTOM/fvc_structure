````python
import numpy as np, os, glob, pickle

base = '/home/nvidia/projects/fvc_structure/ForestFormer3D/data/ForAINetV2'

# The remap applied at pipeline load time (RemapSemanticMask_ in config)
RAW4_TO_FOREST3 = {0: 0, 1: 0, 2: 1, 3: 2}

def apply_remap(arr, mapping, default=3):
    out = np.full_like(arr, default)
    for k, v in mapping.items():
        out[arr == k] = v
    return out

print("=" * 80)
print("REMAP VERIFICATION: new semantic_mask/ + remap  vs  semantic_mask_ori/")
print("=" * 80)

scenes = sorted([
    os.path.splitext(f)[0]
    for f in os.listdir(f'{base}/semantic_mask_ori')
    if f.endswith('.bin')
])

ok = 0
mismatch = 0
for stem in scenes:
    p_ori = f'{base}/semantic_mask_ori/{stem}.bin'
    p_new = f'{base}/semantic_mask/{stem}.bin'
    if not os.path.exists(p_new):
        print(f"  SKIP (no new file): {stem}")
        continue

    ori = np.fromfile(p_ori, dtype=np.int64)
    new = np.fromfile(p_new, dtype=np.int64)
    remapped = apply_remap(new, RAW4_TO_FOREST3)

    ori_cls  = sorted(set(ori.tolist()))
    remap_cls = sorted(set(remapped.tolist()))

    # Check point counts per class
    match = True
    for cls in range(3):
        if (ori == cls).sum() != (remapped == cls).sum():
            match = False
            break

    status = "✓ MATCH" if match else "✗ MISMATCH"
    if not match:
        mismatch += 1
        print(f"\n{status}: {stem}")
        print(f"  ORI   classes+counts: { {c: int((ori==c).sum()) for c in ori_cls} }")
        print(f"  REMAP classes+counts: { {c: int((remapped==c).sum()) for c in remap_cls} }")
    else:
        ok += 1

print(f"\nResult: {ok} scenes MATCH, {mismatch} MISMATCH")
if mismatch == 0:
    print("✓ RemapSemanticMask_ perfectly recreates the original FOREST3 labels for all training scenes.")
else:
    print("⚠ Discrepancies found — the remap does NOT perfectly recreate training labels.")
    print("  This means training with the remap will see DIFFERENT labels than the checkpoint was trained on.")
````

## Semantic Labels Analysis

The semantic labels are stored separately in binary files. Each scene has a `pts_semantic_mask_path` that points to the semantic label file.

````python
import pickle
import numpy as np
from collections import Counter
import os

# Load PKL file
# file_path = '/home/nvidia/projects/fvc_structure/ForestFormer3D/data/ForAINetV2/forainetv2_oneformer3d_infos_train.pkl'
file_path = '/home/nvidia/projects/fvc_structure/ForestFormer3D/data/ForAINetV2/forainetv2_oneformer3d_infos_test.pkl'
data_dir = '/home/nvidia/projects/fvc_structure/ForestFormer3D/data/ForAINetV2/'

with open(file_path, 'rb') as f:
    data = pickle.load(f)

data_list = data.get('data_list', [])

# Semantic class mapping
semantic_mapping = {0: 'ground', 1: 'wood', 2: 'leaf'}

print("=" * 80)
print("SEMANTIC LABELS ANALYSIS")
print("=" * 80)

# Analyze first scene as example
first_scene = data_list[0]
scene_name = first_scene.get('lidar_points', {}).get('lidar_path', 'N/A')
semantic_mask_path = first_scene.get('pts_semantic_mask_path', 'N/A')
instance_mask_path = first_scene.get('pts_instance_mask_path', 'N/A')

print(f"\nExample Scene: {scene_name}")
print(f"Semantic Mask Path (from PKL): {semantic_mask_path}")
print(f"Instance Mask Path (from PKL): {instance_mask_path}")

# Load semantic mask file - handle path with or without subdirectory
if semantic_mask_path != 'N/A':
    # Check if path already includes subdirectory
    if not semantic_mask_path.startswith('semantic_mask/'):
        # Path is just filename, add subdirectory
        full_semantic_path = os.path.join(data_dir, 'semantic_mask', semantic_mask_path)
    else:
        full_semantic_path = os.path.join(data_dir, semantic_mask_path)
    
    print(f"Full Semantic Path: {full_semantic_path}")
    print(f"File exists: {os.path.exists(full_semantic_path)}")
    
    if os.path.exists(full_semantic_path):
        semantic_labels = np.fromfile(full_semantic_path, dtype=np.int64)
        
        print(f"\nSemantic Labels Shape: {semantic_labels.shape}")
        print(f"Total Points: {len(semantic_labels):,}")
        
        # Count semantic classes
        unique, counts = np.unique(semantic_labels, return_counts=True)
        
        print(f"\n{'Class ID':<10} | {'Class Name':<12} | {'Point Count':<15} | {'Percentage':<10}")
        print("-" * 60)
        for class_id, count in zip(unique, counts):
            class_name = semantic_mapping.get(class_id, f'Unknown({class_id})')
            percentage = (count / len(semantic_labels)) * 100
            print(f"{class_id:<10} | {class_name:<12} | {count:<15,} | {percentage:>6.2f}%")
    else:
        print("WARNING: Semantic mask file not found!")

# Also load instance mask to see instance IDs
if instance_mask_path != 'N/A':
    # Check if path already includes subdirectory
    if not instance_mask_path.startswith('instance_mask/'):
        full_instance_path = os.path.join(data_dir, 'instance_mask', instance_mask_path)
    else:
        full_instance_path = os.path.join(data_dir, instance_mask_path)
    
    if os.path.exists(full_instance_path):
        instance_labels = np.fromfile(full_instance_path, dtype=np.int64)
        
        unique_instances = np.unique(instance_labels)
        print(f"\nInstance Mask Analysis:")
        print(f"  - Total unique instance IDs: {len(unique_instances)}")
        print(f"  - Instance ID range: {instance_labels.min()} to {instance_labels.max()}")

# Global semantic statistics across all scenes
print("\n" + "=" * 80)
print("GLOBAL SEMANTIC STATISTICS (First 10 Training Scenes)")
print("=" * 80)

global_semantic_counts = Counter()
global_instance_counts = []
scenes_processed = 0

for scene in data_list[:10]:  # Analyze first 10 scenes for speed
    semantic_mask_path = scene.get('pts_semantic_mask_path')
    instance_mask_path = scene.get('pts_instance_mask_path')
    
    if semantic_mask_path:
        # Handle path with or without subdirectory
        if not semantic_mask_path.startswith('semantic_mask/'):
            full_path = os.path.join(data_dir, 'semantic_mask', semantic_mask_path)
        else:
            full_path = os.path.join(data_dir, semantic_mask_path)
        
        try:
            if os.path.exists(full_path):
                sem_labels = np.fromfile(full_path, dtype=np.int64)
                unique, counts = np.unique(sem_labels, return_counts=True)
                for class_id, count in zip(unique, counts):
                    global_semantic_counts[class_id] += count
                
                # Also count instances
                if instance_mask_path:
                    if not instance_mask_path.startswith('instance_mask/'):
                        inst_path = os.path.join(data_dir, 'instance_mask', instance_mask_path)
                    else:
                        inst_path = os.path.join(data_dir, instance_mask_path)
                    
                    if os.path.exists(inst_path):
                        inst_labels = np.fromfile(inst_path, dtype=np.int64)
                        num_instances = len(np.unique(inst_labels))
                        global_instance_counts.append(num_instances)
                
                scenes_processed += 1
        except Exception as e:
            print(f"Error processing scene {semantic_mask_path}: {e}")

print(f"\nProcessed {scenes_processed} scenes")
print(f"Average instances per scene: {np.mean(global_instance_counts):.1f}" if global_instance_counts else "N/A")

print(f"\n{'Class ID':<10} | {'Class Name':<12} | {'Total Points':<15} | {'Percentage':<10}")
print("-" * 60)
total_points = sum(global_semantic_counts.values())
for class_id in sorted(global_semantic_counts.keys()):
    class_name = semantic_mapping.get(class_id, f'Unknown({class_id})')
    count = global_semantic_counts[class_id]
    percentage = (count / total_points * 100) if total_points > 0 else 0
    print(f"{class_id:<10} | {class_name:<12} | {count:<15,} | {percentage:>6.2f}%")
````

## End-to-end label-space sanity (test PLY → semantic_mask/instance_mask → FOREST3)


This block checks that the classes/instances you see match the intended pipeline:



- **PLY (`test_data/*.ply`)** stores a raw `semantic_seg` (often in a compact RAW5-like space) and `treeID` (instance id).

- **Preprocess (`data/ForAINetV2/load_forainetv2_data.py`)** remaps the PLY semantic ids into a stable **RAW4** space and forces `treeID>0 → RAW4=3 (mallee)`.

  - RAW4 is: `0=forb`, `1=hummock`, `2=shrub`, `3=mallee`

- **On disk (`semantic_mask/*.bin`)** stores **RAW4** ids.

- **Model space (FOREST3)** is produced at runtime by `RemapSemanticMask_` in the config:

  - `RAW4_TO_FOREST3 = {0:0, 1:0, 2:1, 3:2}`

  - FOREST3 is: `0=ground (stuff)`, `1=wood (thing)`, `2=leaf (thing)`



Important notes:

- `treeID>0` points should end up as **RAW4=3 → FOREST3=2**, and **only those** should carry non-zero instance ids.

- FOREST3 class `1` is still a **thing** class (semantic), but we only evaluate instances for the pooled tree bucket (FOREST3 `2`) in our custom metric.


````python
import numpy as np
from plyfile import PlyData
import glob
import os
from collections import Counter, defaultdict

print("=" * 80)
print("COMPLETE DATA PIPELINE VERIFICATION")
print("=" * 80)

# Paths (host paths; also valid in-container due to bind mount/symlink)
base = '/home/nvidia/projects/fvc_structure/ForestFormer3D/data/ForAINetV2'
ply_dir = os.path.join(base, 'test_data')
sem_dir = os.path.join(base, 'semantic_mask')
ins_dir = os.path.join(base, 'instance_mask')

IGNORE_INDEX = -100
RAW4_NAMES = {0: 'forb', 1: 'hummock', 2: 'shrub', 3: 'mallee'}
FOREST3_NAMES = {0: 'ground', 1: 'wood', 2: 'leaf'}
RAW4_TO_FOREST3 = {0: 0, 1: 0, 2: 1, 3: 2, -1: IGNORE_INDEX}

# Known raw schemas (for reference)
RAW5_TO_RAW4 = {0: 0, 1: -1, 2: 1, 3: 2, 4: 3}

def ply_sem_to_raw4(semantic_seg: np.ndarray, tree_id: np.ndarray) -> np.ndarray:
    """Match the remap behavior in data/ForAINetV2/load_forainetv2_data.py.

    Target RAW4 IDs:
      0=forb, 1=hummock, 2=shrub, 3=mallee; -1=unlabeled/ignored
    """
    sem = semantic_seg.astype(np.int64, copy=False)
    tid = tree_id.astype(np.int64, copy=False)

    out = np.full(sem.shape, -1, dtype=np.int64)
    uniq = np.unique(sem)
    uniq_set = set(int(v) for v in uniq.tolist())

    def is_subset(values, allowed):
        return all(v in allowed for v in values)

    # Fast-path A: already in 0-indexed target label space
    if 0 in uniq_set and is_subset(uniq_set, {-1, 0, 1, 2, 3}):
        out = sem.copy()
        out[out < 0] = -1
    # Fast-path B: 1-indexed FOREST3 {1,2,3} -> shift to {0,1,2}
    elif is_subset(uniq_set - {-1}, {1, 2, 3}):
        out = sem.astype(np.int64, copy=True)
        out[out >= 0] -= 1
        out[sem < 0] = -1
    else:
        # (a) LAS/classification style codes 64..68
        if is_subset(uniq_set, {64, 65, 66, 67, 68} | {-1}):
            raw_to_target = {64: 0, 66: 1, 67: 2, 68: 3, 65: -1}
            for raw_id, new_id in raw_to_target.items():
                out[sem == raw_id] = new_id
        # (b) compact 0..4: {forb, woody, hummock, shrub, mallee}
        elif is_subset(uniq_set, {0, 1, 2, 3, 4} | {-1}):
            for raw_id, raw4_id in RAW5_TO_RAW4.items():
                out[sem == raw_id] = raw4_id
        # (c) compact 1..5
        elif is_subset(uniq_set, {1, 2, 3, 4, 5} | {-1}):
            sem0 = sem - 1
            raw_to_target = {0: 0, 2: 1, 3: 2, 4: 3, 1: -1}
            for raw_id, new_id in raw_to_target.items():
                out[sem0 == raw_id] = new_id
        else:
            # Fallback: keep non-negative values (still force tid>0)
            out = sem.copy()
            out[out < 0] = -1

    # Always force mallee wherever treeID indicates an instance.
    out[tid > 0] = 3
    out[out < -1] = -1
    return out

def raw4_to_forest3(raw4: np.ndarray) -> np.ndarray:
    out = np.full(raw4.shape, IGNORE_INDEX, dtype=np.int64)
    for k, v in RAW4_TO_FOREST3.items():
        out[raw4 == k] = v
    return out

# -----------------------------------------------------------------------------
# 1. SOURCE PLY FILES (test_data)
ply_files = sorted(glob.glob(os.path.join(ply_dir, 'znorm_3Dlabels_*.ply')))
print(f"\n1. SOURCE PLY FILES ({len(ply_files)} files)")
print("-" * 80)

all_ply_classes = set()
for ply_file in ply_files:
    plydata = PlyData.read(ply_file)
    vertex = plydata['vertex']
    if 'semantic_seg' in vertex.data.dtype.names:
        all_ply_classes.update(np.unique(vertex['semantic_seg']).tolist())
all_ply_classes = sorted(int(v) for v in all_ply_classes)
print(f"All semantic classes found in PLY files: {all_ply_classes}")

print(f"\nInstances per class (first 3 files):")
class_inst_summary = {}
for ply_file in ply_files[:3]:
    plydata = PlyData.read(ply_file)
    vertex = plydata['vertex']
    if not ('semantic_seg' in vertex.data.dtype.names and 'treeID' in vertex.data.dtype.names):
        continue
    sem = vertex['semantic_seg']
    tid = vertex['treeID']
    for cls in np.unique(sem):
        cls = int(cls)
        cls_mask = (sem == cls)
        cls_instances = tid[cls_mask]
        n_inst = len(np.unique(cls_instances[cls_instances > 0]))
        class_inst_summary.setdefault(cls, []).append(int(n_inst))
for cls in sorted(class_inst_summary.keys()):
    total = sum(class_inst_summary[cls])
    avg = float(np.mean(class_inst_summary[cls]))
    print(f"  Class {cls}: {total} total instances, {avg:.1f} avg per file")

# -----------------------------------------------------------------------------
# 2. PREPROCESSED .BIN FILES (semantic_mask/, instance_mask/)
print(f"\n2. PREPROCESSED .BIN FILES")
print("-" * 80)

sem_files = sorted(glob.glob(os.path.join(sem_dir, 'znorm_3Dlabels_*.bin')))
ins_files = sorted(glob.glob(os.path.join(ins_dir, 'znorm_3Dlabels_*.bin')))
print(f"semantic_mask bin files: {len(sem_files)}")
print(f"instance_mask bin files: {len(ins_files)}")

all_bin_classes = set()
for sem_file in sem_files:
    sem = np.fromfile(sem_file, dtype=np.int64)
    all_bin_classes.update(np.unique(sem).tolist())
all_bin_classes = sorted(int(v) for v in all_bin_classes)
print(f"All semantic classes found in .bin files: {all_bin_classes}")

print(f"\nInstances per class (first 3 files):")
class_inst_summary = {}
for sem_file, ins_file in list(zip(sem_files, ins_files))[:3]:
    sem = np.fromfile(sem_file, dtype=np.int64)
    ins = np.fromfile(ins_file, dtype=np.int64)
    for cls in np.unique(sem):
        cls = int(cls)
        cls_mask = (sem == cls)
        cls_instances = ins[cls_mask]
        n_inst = len(np.unique(cls_instances[cls_instances > 0]))
        class_inst_summary.setdefault(cls, []).append(int(n_inst))
for cls in sorted(class_inst_summary.keys()):
    total = sum(class_inst_summary[cls])
    avg = float(np.mean(class_inst_summary[cls]))
    print(f"  Class {cls}: {total} total instances, {avg:.1f} avg per file")

# -----------------------------------------------------------------------------
# 3. MAPPING VERIFICATION (PLY -> RAW4 -> FOREST3)
print(f"\n3. MAPPING VERIFICATION")
print("-" * 80)

# Empirically summarize PLY-id -> RAW4-id outcomes using the same remap as the loader.
# Note: this can vary if different raw schemas are mixed across files.
ply_to_raw4_counts = defaultdict(Counter)
for ply_file in ply_files:
    plydata = PlyData.read(ply_file)
    vertex = plydata['vertex']
    if not ('semantic_seg' in vertex.data.dtype.names and 'treeID' in vertex.data.dtype.names):
        continue
    sem = vertex['semantic_seg'].astype(np.int64)
    tid = vertex['treeID'].astype(np.int64)
    raw4 = ply_sem_to_raw4(sem, tid)
    for ply_cls in np.unique(sem):
        ply_cls = int(ply_cls)
        mask = sem == ply_cls
        ply_to_raw4_counts[ply_cls].update(raw4[mask].tolist())

print("PLY semantic ids → RAW4 ids (semantic_mask/*.bin) → FOREST3 ids (model)")
for ply_cls in sorted(ply_to_raw4_counts.keys()):
    # Most common RAW4 mapping for this PLY id
    most_common = ply_to_raw4_counts[ply_cls].most_common(3)
    best_raw4 = int(most_common[0][0]) if most_common else None
    best_f3 = RAW4_TO_FOREST3.get(best_raw4, IGNORE_INDEX) if best_raw4 is not None else IGNORE_INDEX
    raw4_name = RAW4_NAMES.get(best_raw4, str(best_raw4))
    f3_name = FOREST3_NAMES.get(best_f3, str(best_f3))
    dist = ", ".join([f"{int(k)}:{int(v)}" for k, v in most_common])
    print(f"  PLY {ply_cls:>2} → RAW4 {best_raw4} ({raw4_name}) → FOREST3 {best_f3} ({f3_name})  [top: {dist}]")

print("\nModel semantic classes:")
print("  FOREST3 0 = ground (stuff)")
print("  FOREST3 1 = wood   (thing; semantic)")
print("  FOREST3 2 = leaf   (thing; semantic; carries tree instances after pooling)")

# One-file consistency check (choose a file that actually has instances if available)
if ply_files and sem_files and ins_files:
    chosen_stem = None
    for sem_path, ins_path in zip(sem_files, ins_files):
        ins = np.fromfile(ins_path, dtype=np.int64)
        if (ins > 0).any():
            chosen_stem = os.path.splitext(os.path.basename(sem_path))[0]
            break
    if chosen_stem is None:
        chosen_stem = os.path.splitext(os.path.basename(ply_files[0]))[0]

    ply_path = os.path.join(ply_dir, chosen_stem + '.ply')
    sem_path = os.path.join(sem_dir, chosen_stem + '.bin')
    ins_path = os.path.join(ins_dir, chosen_stem + '.bin')

    plydata = PlyData.read(ply_path)
    vertex = plydata['vertex']
    sem_ply = vertex['semantic_seg'].astype(np.int64)
    tid_ply = vertex['treeID'].astype(np.int64)
    raw4_from_ply = ply_sem_to_raw4(sem_ply, tid_ply)
    forest3_from_ply = raw4_to_forest3(raw4_from_ply)

    sem_bin = np.fromfile(sem_path, dtype=np.int64)
    ins_bin = np.fromfile(ins_path, dtype=np.int64)
    forest3_from_bin = raw4_to_forest3(sem_bin)

    print("\nOne-file check (unique labels):", chosen_stem)
    print("  PLY unique semantic_seg:", sorted(int(v) for v in np.unique(sem_ply).tolist()))
    print("  RAW4 from PLY (after remap + treeID forcing) uniques:", sorted(int(v) for v in np.unique(raw4_from_ply).tolist()))
    print("  RAW4 from semantic_mask/*.bin uniques:", sorted(int(v) for v in np.unique(sem_bin).tolist()))
    print("  FOREST3 from RAW4 (.bin) uniques:", sorted(int(v) for v in np.unique(forest3_from_bin).tolist()))
    # Instance sanity: non-zero instance ids should be on RAW4==3 only
    inst_nonzero = ins_bin > 0
    if inst_nonzero.any():
        raw4_where_inst = np.unique(sem_bin[inst_nonzero])
        print("  RAW4 labels where instance_id>0:", sorted(int(v) for v in raw4_where_inst.tolist()))
    else:
        print("  No instance_id>0 found in this file (this can be normal if there are no mallee instances).")

print("\n" + "=" * 80)
````

## PLY Structure Comparison: train_val_data vs test_data

````python
import numpy as np
import glob
import os
from plyfile import PlyData
from collections import Counter

TRAIN_DIR = "/home/nvidia/projects/fvc_structure/ForestFormer3D/data/ForAINetV2/train_val_data"
TEST_DIR  = "/home/nvidia/projects/fvc_structure/ForestFormer3D/data/ForAINetV2/test_data"

# Label mappings
TRAIN_CLS = {0: "ground", 1: "wood", 2: "leaf"}          # FOREST3 space
TEST_CLS  = {2: "hummock", 3: "shrub", 4: "mallee"}       # RAW5 on-disk classes

def safe_read_ply(fp):
    """Read PLY, falling back to non-mmap mode; return None on failure."""
    try:
        return PlyData.read(fp, mmap=False)
    except Exception as e:
        print(f"    [SKIP] {os.path.basename(fp)}: {e}")
        return None

def summarise_dir(directory, label_map, sem_field="semantic_seg", inst_field="treeID", n_sample=3):
    ply_files = sorted(glob.glob(os.path.join(directory, "*.ply")))
    print(f"  Directory : {directory}")
    print(f"  PLY count : {len(ply_files)} files")

    # Find a readable sample file for schema
    sample_ply = None
    sample_path = None
    for fp in ply_files:
        tmp = safe_read_ply(fp)
        if tmp is not None:
            sample_ply = tmp
            sample_path = fp
            break

    if sample_ply is None:
        print("  ERROR: no readable PLY found"); return

    vertex = sample_ply["vertex"]
    props = [p.name for p in vertex.properties]
    print(f"  Sample    : {os.path.basename(sample_path)}")
    print(f"  Properties: {props}")
    print(f"  N_points  : {vertex.count:,}")
    print()

    # Aggregate class + instance stats across first n_sample readable files
    class_totals = Counter()
    total_pts_all = 0
    sample_results = []   # (basename, n_instances) per file
    sampled = 0

    for fp in ply_files:
        if sampled >= n_sample:
            break
        ply = safe_read_ply(fp)
        if ply is None:
            continue
        v = ply["vertex"]
        total_pts_all += v.count
        sampled += 1
        if sem_field in v.data.dtype.names:
            sem = v[sem_field]
            u, c = np.unique(sem, return_counts=True)
            for cls_id, cnt in zip(u, c):
                class_totals[int(cls_id)] += int(cnt)
        n_inst = 0
        if inst_field in v.data.dtype.names:
            inst = v[inst_field]
            n_inst = len(np.unique(inst[inst > 0]))
        sample_results.append((os.path.basename(fp), n_inst))

    print(f"  --- Semantic class distribution (first {sampled} readable files) ---")
    print(f"  {'Class ID':<10} {'Name':<12} {'Points':>12}")
    print(f"  {'-'*38}")
    for cls_id in sorted(class_totals):
        name = label_map.get(cls_id, f"unknown({cls_id})")
        pts  = class_totals[cls_id]
        print(f"  {cls_id:<10} {name:<12} {pts:>12,}")

    if sample_results:
        print(f"\n  --- Instance counts (first {sampled} readable files) ---")
        for fname, ni in sample_results:
            print(f"    {fname}: {ni} instances")

    # Deeper: per-class instance check on the sample file
    v = sample_ply["vertex"]
    if sem_field in v.data.dtype.names and inst_field in v.data.dtype.names:
        sem = v[sem_field]; inst = v[inst_field]
        print(f"\n  --- Per-class instance check (sample: {os.path.basename(sample_path)}) ---")
        for cls_id in np.unique(sem):
            mask = (sem == cls_id)
            ni = len(np.unique(inst[mask & (inst > 0)]))
            name = label_map.get(int(cls_id), f"unknown({cls_id})")
            tag = "✓ HAS instances" if ni > 0 else "  no instances"
            print(f"    class {cls_id} ({name:<10}): {tag}  ({ni} unique IDs)")

# ─────────────────────────────────────────────────────────────
print("=" * 70)
print("TRAIN/VAL DATA  —  FOREST3 label space (ground / wood / leaf)")
print("=" * 70)
summarise_dir(TRAIN_DIR, TRAIN_CLS)

print()
print("=" * 70)
print("TEST DATA  —  RAW5 on-disk label space (hummock / shrub / mallee)")
print("=" * 70)
summarise_dir(TEST_DIR, TEST_CLS)

print()
print("=" * 70)
print("KEY DIFFERENCES SUMMARY")
print("=" * 70)
rows = [
    ("Naming scheme",       "CORPUS_site_plotN_{split}.ply",  "znorm_3Dlabels_{type}_{id}.ply"),
    ("Label space",         "FOREST3: {0,1,2}",               "RAW5: {2,3,4}"),
    ("Semantic classes",    "0=ground, 1=wood, 2=leaf",        "2=hummock, 3=shrub, 4=mallee"),
    ("Tree instances",      "cls 1 (wood) + cls 2 (leaf)",     "cls 4 (mallee) only"),
    ("Corpus origin",       "CULS / NIBIO / BlueCat (Norway)", "Calperum (Australia)"),
    ("Remap at model load", "none — already FOREST3",          "RemapSemanticMask_ 2→0,3→1,4→2"),
]
print(f"  {'Attribute':<25} {'train_val_data':<35} {'test_data'}")
print(f"  {'-'*95}")
for attr, train_val, test in rows:
    print(f"  {attr:<25} {train_val:<35} {test}")
````
